A formula de Herão permite que a pessoa calcule a área sem precisar da altura exata do triângulo. Nós aprendedemos que
para achar a área precisamos da altura e da base então multiplicamos e depois dividimos por 2, mas podemos usar a seguinte
fórmula

![Formula de Herão](Herao.png)

Com s sendo o semiperimetro dado por:

![Fórmula do semiperimetro](Semiperimetro.png)

```python
from math import sqrt

lados = []

while True:
    try:
        lado_a = float(input("Digite o valor do lado A: "))
        lados.append(lado_a)
        lado_b = float(input("Digite o valor do lado B: "))
        lados.append(lado_b)
        lado_c = float(input("Digite o valor do lado C: "))
        lados.append(lado_c)
        break
    except ValueError:
        print("Valor invalido")

s = sum(lados)/2

area = sqrt(s*(s - lado_a)*(s - lado_b)*(s - lado_c))

print(f"Area: {area:.2f}")
```
